```python
import pandas as pd
import numpy as np
import joblib
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# --- Load validation CSV ---
val_df = pd.read_csv("C:/Users/swaya/Downloads/val.csv")

# Separate features and target
X_val = val_df.drop(["url", "label"], axis=1)
y_val = val_df["label"]

print("Validation shape:", X_val.shape)
```

```python
# Detect object/string columns
obj_cols = X_val.select_dtypes(include='object').columns
print("Encoding object columns:", list(obj_cols))

# Encode each using LabelEncoder (temporary solution)
for col in obj_cols:
    le = LabelEncoder()
    X_val[col] = le.fit_transform(X_val[col].astype(str))
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier()
xgb_model.load_model("C:/Users/swaya/Desktop/project/urlying/ipynbs/training/models/xgboost_model_with_no_feature_filtration.json")

# Predict
y_pred_xgb = xgb_model.predict(X_val)

print("XGBoost Results:")
print("Accuracy:", accuracy_score(y_val, y_pred_xgb))
print("Precision:", precision_score(y_val, y_pred_xgb, average='weighted'))
print("Recall:", recall_score(y_val, y_pred_xgb, average='weighted'))
print("F1:", f1_score(y_val, y_pred_xgb, average='weighted'))
```

```python
# Load MLP model and its scaler
mlp_model = joblib.load("C:/Users/swaya/Desktop/project/urlying/ipynbs/training/models/mlp_model.pkl")
mlp_scaler = joblib.load("C:/Users/swaya/Desktop/project/urlying/ipynbs/training/models/mlp_scalar.pkl")

# Scale features
X_val_scaled = mlp_scaler.transform(X_val)

# Predict
y_pred_mlp = mlp_model.predict(X_val_scaled)

print("MLP Results:")
print("Accuracy:", accuracy_score(y_val, y_pred_mlp))
print("Precision:", precision_score(y_val, y_pred_mlp, average='weighted'))
print("Recall:", recall_score(y_val, y_pred_mlp, average='weighted'))
print("F1:", f1_score(y_val, y_pred_mlp, average='weighted'))
```
